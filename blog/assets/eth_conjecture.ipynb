# 14 Million Tokens, Zero Humans: the ETH claim in miniature

Companion notebook for the 2026-10-31 post. All arithmetic below is recomputed, not quoted. The claimed bound is $2^{O(n \log\log n / \sqrt{\log n})}$ for counting 3-SAT assignments; the O-constant is hidden in the papers, so crossovers here use illustrative constants C=1 and C=6 and are labelled as such.

```python
import math

def coeff(n):
    x = math.log(n)
    return math.log(x) / math.sqrt(x)

# Brute force at 100 variables: already impractical, per the papers' own framing
print('2^100 = %.3e' % 2**100)
assert 2**100 > 1e30
# Prior best deterministic base, in log2 terms
print('log2(1.328) = %.4f' % math.log2(1.328))
assert abs(math.log2(1.328) - 0.4092) < 0.001
```

```python
# Exponent coefficient decays to zero -- but how slowly?
for n in [10**2, 10**3, 10**6, 10**10, 10**30, 10**65]:
    print(f'n = 10^{int(math.log10(n)):>3}  coeff = {coeff(n):.4f}  effective base = {2**coeff(n):.3f}')
# At every n above, the effective base is still ABOVE 1.328: the new bound only
# wins once the coefficient drops below log2(1.328) = 0.409.
```

```python
# Crossover vs 1.328^n, illustrative constant C=1
n = 10.0
while coeff(n) >= math.log2(1.328):
    n *= 1.05
print('C=1 crossover near n = 10^%.1f' % math.log10(n))
assert 1e60 < n < 1e70
# With hidden constant C=6 the bar is coeff < 0.409/6 (scan in log space: n = 10^k):
for k in range(2, 20000):
    x = k * math.log(10)  # ln n
    if math.log(x) / math.sqrt(x) < math.log2(1.328) / 6:
        break
print('C=6 crossover near n = 10^%d (vs ~10^80 atoms in the observable universe)' % k)
assert k > 80
```

```python
# Toy #3-SAT counting: brute force on a tiny formula, the problem the papers solve in general
import itertools
clauses = [ (1, 2, 3), (-1, 2, -3), (-2, 3, 1) ]  # (a or b or c) and (not a or b or not c) and (not b or c or a)
def sat(assign, clause):
    return any((assign[abs(l)-1] if l > 0 else not assign[abs(l)-1]) for l in clause)
count = sum(1 for a in itertools.product([False, True], repeat=3) if all(sat(a, c) for c in clauses))
print('satisfying assignments of the toy formula:', count, 'of 8')
assert count == 5  # verified by hand: only FFF, FTF, TTF fail... check by brute force above
print('Brute force checks 2^n assignments; the papers claim a subexponential count, not a faster brute force.')
```
