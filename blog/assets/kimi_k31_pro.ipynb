Recomputes every headline figure from the launch coverage (Oct 23-27, 2026): the Intelligence Index jump, the cyber cut, the price math, and the verbosity trap that eats the discount.

```python
# Intelligence Index: 44 -> 55, rank #4 of 215
k3, k31 = 44, 55
print(f"gain: +{k31-k3} points, +{(k31-k3)/k3:.1%} relative")
assert k31-k3 == 11
# Gap to the frontier: Opus 5.5 at 58, Sonnet 5.5 at 56
print(f"gap to Opus: {58-k31} points (was {58-k3} for K3) — the open-weight gap shrank 14 -> 3")
assert 58-k31 == 3 and 58-k3 == 14
```

```python
# The cyber cut, xdev-bench-hard pass@3 (Moonshot's own test)
uncut, shipped, k3_old, rival = 90.5, 81.0, 33.3, 85.7
print(f"cut: -{uncut-shipped:.1f}pp ({(uncut-shipped)/uncut:.1%} relative)")
print(f"shipped vs predecessor: {shipped/k3_old:.2f}x — the cut is a dial, not a wall")
print(f"shipped vs DeepSeek V4.1 Flash: {shipped-rival:+.1f}pp (below it, not above)")
assert uncut-shipped == 9.5 and abs(shipped/k3_old-2.43) < 0.01
```

```python
# Agentic gains: GDPval-AA and AutomationBench
print(f"GDPval-AA: 1534 -> 1846 (+{1846-1534}, +{(1846-1534)/1534:.1%})")
print(f"AutomationBench: 58% -> 73% (+15pp)")
print(f"Hard hallucination (Moonshot internal): 23.4% -> 8.3% (-15.1pp, -64.5% relative)")
assert 1846-1534 == 312 and abs((23.4-8.3)/23.4-0.645) < 0.01
# Terminal-Bench 4.0 harness warning: Moonshot's table 65.7% vs Opus 66.4%;
# an independent run reported 67.0%. Same model, different harness, different headline.
print("Terminal-Bench 4.0: Moonshot 65.7% vs independent 67.0% — harness is a variable, not a detail")
```

```python
# Price math — and the verbosity trap
def task_cost(inp_tokens, out_tokens, pin, pout):
    return inp_tokens/1e6*pin + out_tokens/1e6*pout
# Toy task: 50k input tokens, 100k output tokens
opus  = task_cost(50_000, 100_000, 4, 10*2)   # $4/$20
k31   = task_cost(50_000, 100_000, 2, 10)    # $2/$10
k31_v = task_cost(50_000, 150_000, 2, 10)    # 1.5x output verbosity
print(f"Opus task: ${opus:.2f} | K3.1 equal verbosity: ${k31:.2f} (exactly half)")
print(f"K3.1 at 1.5x verbosity: ${k31_v:.2f} (only {1-k31_v/opus:.0%} cheaper — verbosity eats the discount)")
assert opus == 2.20 and k31 == 1.10 and k31_v == 1.60
# K3 price cut: $3/$15 -> $2/$10 is exactly one third off
assert abs((3-2)/3 - 1/3) < 1e-9 and abs((15-10)/15 - 1/3) < 1e-9
print("K3 -> K3.1 list price: -33.3% on both input and output")
```
